````python
students = {
    "Alice": [85, 90, 78],
    "Bob": [70, 88, 92],
    "Charlie": [95, 100, 93],
    "Diana": [60, 75, 72]
}

### 1. Write a Function to Calculate the Average
#calculate_average(grades)
##Requirements:
# Accepts a list of numbers.
# Returns the average.
# Must not print inside the function.

def calculate_average(grades):
    if not grades:
        return 0
    return sum(grades) / len(grades)

for name, grades in students.items():
    print(f"{name}: {calculate_average(grades):.2f}")
````

````python
##2. Write a Function to Assign a Letter Grade
# assign_letter_grade(average, passing_mark=60)

#Requirements:
# Uses a default parameter `passing_mark=60`.
# Returns:
# A if average ≥ 90
# B if average ≥ 80
# C if average ≥ 70
# D if average ≥ passing_mark
# F otherwise

def assign_letter_grade(average, passing_mark=60):
    if average >= 90:
        return 'A'
    elif average >= 80:
        return 'B'
    elif average >= 70:
        return 'C'
    elif average >= passing_mark:
        return 'D'
    else:
        return 'F'
    
print("\nLetter Grades:")
for name, grades in students.items():
    average = calculate_average(grades)
    letter_grade = assign_letter_grade(average)
    print(f"{name}: {letter_grade}")    
````

### **4.Structure Requirements**

````python
expenses = {
    "January": [120, 340, 560],
    "February": [400, 150],
    "March": [800, 200, 100]
}

### **1. Monthly Total Function**

# monthly_total(expense_list)
# Receives a list of expenses.
# Calculates and returns the total expenses for a month.
# Use a `for` loop to calculate the total.

def monthly_total(expense_list):
    total = 0
    for expense in expense_list:
        total += expense
    return total
for month, expenses_list in expenses.items():
    print(f"{month}: {monthly_total(expenses_list)}")
````

````python
## 2. Highest Spending Month**

# highest_month(expense_dict)
# Receives the dictionary of monthly expenses.
# Calculates the total expenses for each month using the `monthly_total()` function.
# Finds and returns the name of the month with the highest total spending.
# Hint: You can use the built-in `max()` function to identify the highest monthly total. Think about how you can associate each total with its corresponding month.

def highest_month(Expense_dict):
    highest_month = None
    highest_total = 0
    for month, expenses_list in Expense_dict.items():
        total = monthly_total(expenses_list)
        if total > highest_total:
            highest_total = total
            highest_month = month
    return highest_month
print(f"Highest Spending Month: {highest_month(expenses)}")
````

````python
## 3. Expense Filter
# filter_expenses(expense_list, threshold=300)
# Receives a list of expenses and a threshold.
# Returns expenses greater than the threshold.
# Use list comprehension.

def filter_expenses(expense_list, threshold=300):
    return [expense for expense in expense_list if expense > threshold]

print("\nFiltered Expenses:")
for month, expenses_list in expenses.items():
    filtered = filter_expenses(expenses_list)
    print(f"{month}: {filtered}")
````

````python
## 4.Structure Requirements

# Create a `main()` function.
# Print the total expenses for each month.
# Print the month with the highest spending.
# Print the filtered expenses for each month.
# Call the appropriate functions to perform the required operations.
# Do not use global variables inside functions. Pass the required data as function arguments.
# Use only concepts covered in class: functions, loops, conditional statements, dictionaries, lists, list comprehension, and built-in functions.

def main():
    expenses = {
        "January": [120, 340, 560],
        "February": [400, 150],
        "March": [800, 200, 100]
    }

    def monthly_total(expense_list):
        total = 0
        for expense in expense_list:
            total += expense
        return total

    def highest_month(expense_dict):
        highest_month = None
        highest_total = 0
        for month, expenses_list in expense_dict.items():
            total = monthly_total(expenses_list)
            if total > highest_total:
                highest_total = total
                highest_month = month
        return highest_month

    def filter_expenses(expense_list, threshold=300):
        return [expense for expense in expense_list if expense > threshold]

    print("Monthly Totals:")
    for month, expenses_list in expenses.items():
        print(f"{month}: {monthly_total(expenses_list)}")

    print(f"\nHighest Spending Month: {highest_month(expenses)}")

    print("\nFiltered Expenses:")
    for month, expenses_list in expenses.items():
        filtered = filter_expenses(expenses_list)
        print(f"{month}: {filtered}")
        
main()
````

# **Problem 3 (Inventory Restocking System)**

## Scenario
A store maintains product stock and the minimum required quantities.

```python
inventory = {
    "Apples": {"quantity": 20, "min_required": 25},
    "Bananas": {"quantity": 40, "min_required": 30},
    "Oranges": {"quantity": 15, "min_required": 20}
}
```
## Tasks

### **1. Restock Check**
Create:

```python
needs_restock(product_info)
```

- Returns True if quantity < min_required.



### **2. Restock Amount**
Create:

```python
restock_amount(product_info)
```

- Returns the number of items required to reach minimum stock.



### **3. Generate Restock Report**
Create:

```python
generate_restock_report(inventory)
```

- Returns dictionary of products needing restock:

```python
{"Apples": 5, "Oranges": 5}
```

### **4. Sorting**
- Sort report by highest restock amount.
- Use lambda expression inside `sorted()`.

  **Hint:** Use `sorted()` with a `lambda` to sort by values in a dictionary:

```python
# Example syntax
sorted(restock_report.items(), key=lambda x: x[1], reverse=True)
```

- `x[1]` → accesses the restock amount (value) of each `(product, amount)` pair
- `reverse=True` → sorts from highest to lowest
- The result will be a **list of tuples** in descending restock order
---

````python
# Problem 3 (Inventory Restocking System)
inventory = {
    "Apples": {"quantity": 20, "min_required": 25},
    "Bananas": {"quantity": 40, "min_required": 30},
    "Oranges": {"quantity": 15, "min_required": 20}
}

## 1. Restock Check
# needs_restock(product_info)
# Returns True if quantity < min_required.

def needs_restock(product_info):
    return product_info["quantity"] < product_info["min_required"]
for product, info in inventory.items():
    if needs_restock(info):
        print(f"{product}: TRUE")
    else:
        print(f"{product}: FALSE")
## 2. Restock Amount
# restock_amount(product_info)
# Returns the number of items required to reach minimum stock.

def restock_amount(product_info):
    return max(0, product_info["min_required"] - product_info["quantity"])  

for product, info in inventory.items():
    amount_needed = restock_amount(info)
    print(f"\n{product}: {amount_needed} items needed for restock.")

### 3. Generate Restock Report
# generate_restock_report(inventory)
# Returns dictionary of products needing restock:

def generate_restock_report(inventory):
    report = {}
    for product, info in inventory.items():
        if needs_restock(info):
            report[product] = restock_amount(info)
    return report

print("\nRestock Report:", generate_restock_report(inventory))
    
## 4. Sorting
# Sort report by highest restock amount.
# Use lambda expression inside `sorted()`.
# Hint: Use `sorted()` with a `lambda` to sort by values in a dictionary:

## sorted(restock_report.items(), key=lambda x: x[1], reverse=True)
# `x[1]` → accesses the restock amount (value) of each `(product, amount)` pair
# `reverse=True` → sorts from highest to lowest
# The result will be a **list of tuples** in descending restock order


restock_report = generate_restock_report(inventory)
sorted_report = sorted(restock_report.items(), key=lambda x: x[1], reverse=True)
    
print("\nSorted Restock Report:", sorted_report)
````

# **Problem 4 (Voting Analysis with Sets)**

## Scenario
Two polling stations collected votes:

```python
station_A = {"Alice", "Bob", "Charlie", "Diana"}
station_B = {"Charlie", "Diana", "Edward"}
```

## Tasks

### **1. Common Voters**
Create:

```python
common_voters(A, B)
```

- Return voters appearing in both sets.



### **2. Unique Voters**
Create:

```python
unique_voters(A, B)
```

- Return voters appearing in only one set.



### **3. All Voters**
Create:

```python
all_voters(A, B)
```

- Return union of voters.



### **4. Turnout Analysis**
Create:

```python
analyse_turnout(A, B)
```

- Return dictionary:

```python
{
    "total": ...,
    "common": ...,
    "unique": ...
}
```



### **5. Sorting**
- Sort voters alphabetically using `sorted()` and lambda.

````python
# Problem 4 (Voting Analysis with Sets)

station_A = {"Alice", "Bob", "Charlie", "Diana"}
station_B = {"Charlie", "Diana", "Edward"}

### **1. Common Voters**
## common_voters(A, B)
# Return voters appearing in both sets.

def common_voters(A, B):
    return A.intersection(B)

print(common_voters(station_A, station_B))
### **2. Unique Voters**
## unique_voters(A, B)
# Return voters appearing in only one set.

def unique_voters(A, B):
    return A.symmetric_difference(B)

print(unique_voters(station_A, station_B))

### **3. All Voters**
## all_voters(A, B)
# Return union of voters.


def all_voters(A, B):
    return A.union(B)
print(all_voters(station_A, station_B))

### **4. Turnout Analysis**
## analyse_turnout(A, B)
## Return dictionary:## {
##    "total": ...,
##    "common": ...,
##    "unique": ...
##}

def analyse_turnout(A, B):
    return {
        "total": len(A.union(B)),
        "common": len(A.intersection(B)),
        "unique": len(A.symmetric_difference(B))
    }
print(analyse_turnout(station_A, station_B))

### **5. Sorting**
# Sort voters alphabetically using `sorted()` and lambda.

sorted_voters = sorted(all_voters(station_A, station_B), key=lambda x: x)
print(sorted_voters)
````
