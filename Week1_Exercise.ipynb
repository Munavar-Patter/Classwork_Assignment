## **Problem 7: Type Casting (String → Numeric)**


## **Problem 11: Variable Reassignment**

   - Why did the `id(x)` change?

## **Problem 14: Student Enrollment**
1. Create `age = "19"` (as a string)

## **Problem 15: Mixed Calculation Challenge**
