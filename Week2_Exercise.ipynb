## **Problem A1 (Conditional number categorisation)**

Ask the user to enter a number. 

Print:
- "Positive even" if the number is positive and even
- "Positive odd" if positive and odd
- "Negative even" if negative and even
- "Negative odd" if negative and odd
- "Zero" if the number is 0

````python
X = int(input("Enter a Number:").strip())

if X>0:
    if X%2 == 0:
        print("The number is positive and even.")
    else:
        print("The number is positive and odd.")
elif X<0:
    if X%2 == 0:
        print("The number is negative and even.")
    else:

        print("The number is negative and odd.")
else:
    print("The number is zero.")
````

Ask the user to enter three integers: `a`, `b`, and `c`.

**Task:** print `Valid` if all the following conditions hold:
- `a < b < c`
- the difference between consecutive numbers is equal
- none of the numbers is negative

Otherwise, print `Invalid`.

### Reflection questions

- Why is it useful to validate all constraints explicitly in one problem?


````python
a = int(input("Enter a Number:").strip())
b = int(input("Enter a Number:").strip())
c = int(input("Enter a Number:").strip())

if a<b<c and b-a==c-b and a>=0 and b>=0 and c>=0:
    print("Valid")
else:
    print("Invalid")
````

# **Section B: Loops and Control Flow**

## **Problem B1 (For loop with conditional accumulation)**

Ask the user to enter a positive integer `n`.

Using a `for` loop, calculate the sum of all numbers from `1` to `n` that are divisible by `3`.

Print the final sum.

### Reflection
- How can a loop be used to accumulate only selected values?
- What is the role of the conditional statement inside the loop?

````python
n = int(input("Enter a Number:").strip())
sum = 0
for i in range(1,n+1):
    if i%3==0:
        sum+=i
print(sum)
````

## **Problem B2 (Nested loops with multiplication table)**

Ask the user to enter a positive integer `n`.

Using nested loops, print a multiplication table from `1` to `n`.

Example for `n = 4`:

1 2 3 4  
2 4 6 8  
3 6 9 12  
4 8 12 16

### Reflection
- How do nested loops help generate pairwise combinations?
- Why is this problem difficult to solve with a single loop?


````python
n = int(input("Enter a Number:").strip())
for i in range(1,n+1):
    for j in range(1,n+1):
        print(i*j,end=" ")
    print()
````

## **Problem B3  (Loop control statements: continue and break)**

Ask the user to enter a positive integer `n`.

Using a `for` loop:
- skip numbers divisible by `4`
- stop the loop completely if the number reaches `13`
- print all other numbers from `1` to `n`

### Reflection
- What is the difference between `continue` and `break`?
- How does each statement affect the flow of the loop?

````python
n = int(input("Enter a Number:").strip())
for i in range(1,n+1):
    if i%4==0:
        continue
    if i==13:
        break
    print(i)
````

## **Problem B4 (While loop with input validation)**

Keep asking the user to enter a number until they enter a negative number.

Using a `while` loop:
- count how many non-negative numbers were entered
- compute their total sum

When the user enters a negative number, stop the loop and print:
- the count
- the sum

### Reflection
- Why is a `while` loop more suitable here than a `for` loop?
- What condition controls the repetition in this task?

````python
N = int(input("Enter a Number:").strip())
count = 0
sum = 0
while N>=0:
    count+=1
    sum+=N
    N = int(input("Enter a Number:").strip())
print(count)
print(sum)
````

Create a list of squares for numbers from 1 to 12 using a list comprehension. Print the resulting list.

````python
Numbers = [x for x in range(1,13)]
print(Numbers)
````

## **Problem C2 (Conditional label)**

Using a list comprehension with a conditional expression, generate a list labelling numbers 1 to 15 as "even" or "odd" depending on whether the number is even or odd.

````python
Numbers = ["Even" if x%2==0 else "Odd" for x in range(1,16)]
print(Numbers)
````

## **Problem C3 (Nested list sums)** 

Given two lists `a = [2, 4]` and `b = [5, 10]`, use a nested list comprehension to create a list of all possible sums of one element from `a` and one element from `b`. Print the resulting list.

````python
a = [2,4]
b = [5,10]
Sum = [x+y for x in a for y in b]
print(Sum)
````

# Final Mini Challenge: Smart Student Performance Analyser 
You are given a dataset of students, where each student has a name, a score, and an attendance percentage.
### Given Data
```
students = [
    {"name": "Ali", "score": 85, "attendance": 92},
    {"name": "Sara", "score": 58, "attendance": 75},
    {"name": "Lina", "score": 73, "attendance": 88},
    {"name": "Omar", "score": 49, "attendance": 95},
    {"name": "Noor", "score": 91, "attendance": 98},
    {"name": "Yara", "score": 66, "attendance": 60}
]
```
### Tasks

#### Part 1: Classify each student
- if score ≥ 70 and attendance ≥ 80 → "Good standing"  
- if score < 70 and attendance ≥ 80 → "Needs academic improvement"  
- if score ≥ 70 and attendance < 80 → "Needs attendance improvement"  
- otherwise → "At risk"  

#### Part 2: Count Categories
Print the total number of students in each category

#### Part 3: Warning System
Print warnings if:
- student is "At risk"  
- score < 50  
- attendance < 65  

#### Part 4: List Comprehension
Create a list of students in "Good standing"



````python
students = [
    {"name": "Ali", "score": 85, "attendance": 92},
    {"name": "Sara", "score": 58, "attendance": 75},
    {"name": "Lina", "score": 73, "attendance": 88},
    {"name": "Omar", "score": 49, "attendance": 95},
    {"name": "Noor", "score": 91, "attendance": 98},
    {"name": "Yara", "score": 66, "attendance": 60}
]
## 1. CATEGORIZE STUDENTS
for i in students:
    name = i["name"]
    score = i["score"]
    attendance = i["attendance"]
    if score >= 70 and attendance >= 80:
        i["category"] = "Good Standing"
    elif score < 70 and attendance >= 80:
        i["category"] = "Need academic improvement"
    elif score >= 70 and attendance < 80:
        i["category"] = "Need attendance improvement"
    else:
        i["category"] = "At Risk"
print(students)

## 2. COUNT CATEGORIES
categories = ["Good Standing", "Need academic improvement", "Need attendance improvement", "At Risk"]

for category in categories:
    count = 0
    for student in students:
        if student["category"] == category:
            count += 1
    print(f"{category}: {count}")

## 3. warning system

for student in students:
    warnings = []
    if student["category"] == "At Risk":
        warnings.append("At Risk")
    if student["score"] < 50:
        warnings.append("Low Score")   
    if student["attendance"] < 60:
        warnings.append("Low Attendance")
    if warnings:
        print(student['name'],":",", ".join(warnings))

## Good students
Good_students = [student["name"] for student in students if student["category"] == "Good Standing"]
print(Good_students)
````
