## **Problem A1 (Indexing and slicing)**

Given the list:

```python
temperatures = [18, 21, 19, 25, 30, 28, 22]
```

Print:
- the first and the last temperature (the last one using a **negative index**)
- the three temperatures in the middle of the list (positions 2 to 4)
- every temperature **except** the first two
- the whole list in reverse order using slicing

### Reflection
- Why does `temperatures[1:3]` return only two elements?
- When is a negative index more convenient than a positive one?

````python
temperatures = [18, 21, 19, 25, 30, 28, 22]

# the first and the last temperature (the last one using a negative index)
print(temperatures[0], temperatures[-1])

# the three temperatures in the middle of the list (positions 2 to 4)
print(temperatures[1:4])  

# every temperature except the first two
print(temperatures[2:])

# the whole list in reverse order using slicing
print(temperatures[::-1])
````

1. Why does `temperatures[1:3]` return only two elements?
-- Because the slice includes the element at index 1 but excludes the element at index 3.

2. When is a negative index more convenient than a positive one?
-- A negative index is more convenient when you want to access elements from the end of the list without needing to calculate their positive indices.

## **Problem A3 (Searching, counting and sorting)**

Given the list:

```python
scores = [70, 85, 70, 92, 60, 70, 85]
```

Tasks:
- Print how many times `70` appears
- Print the index of the **first** `85`
- Print `True` or `False` depending on whether `100` is in the list
- Create a **copy** of the list, sort the copy in descending order, and print both the original and the sorted copy

### Reflection
- What would happen to `scores` if you wrote `sorted_scores = scores` instead of using a copy and then sorted `sorted_scores`?

````python
scores = [70, 85, 70, 92, 60, 70, 85]

# Print how many times `70` appears
print(scores.count(70))

# Print the index of the **first** `85`
print(scores.index(85))

# Print `True` or `False` depending on whether `100` is in the list
print(100 in scores)

#Create a **copy** of the list, sort the copy in descending order, and print both the original and the sorted copy
sorted_scores = scores.copy()
sorted_scores.sort(reverse=True)
print("Original list:", scores)
print("Sorted list:", sorted_scores)

````

2. Why does `sort()` return `None`?
-- The `sort()` method sorts the list in place and returns `None` because it modifies the original list directly, rather than returning a new sorted list.

## **Problem A4 (Joining lists and list comprehensions)**

Given:

```python
morning = ["Ali", "Sara", "Lina"]
afternoon = ["Omar", "Noor"]
```

Tasks:
- Create `all_students` by joining the two lists using `+`
- Create a second list by adding the afternoon students to a **copy** of `morning` using `extend()`
- Using a list comprehension, create a list of the numbers from `1` to `20` that are **odd**, each one squared
- Using a list comprehension with a conditional expression, create a list of the numbers from `1` to `10` where even numbers are replaced by `"even"` and odd numbers stay as they are

### Reflection
- What is the difference between `+` and `extend()`?
- Where does the `if` go in a comprehension that *filters*, and where does it go in one that *transforms*?

````python
morning = ["Ali", "Sara", "Lina"]
afternoon = ["Omar", "Noor"]

# Create `all_students` by joining the two lists using `+`
all_students = morning + afternoon
print(all_students)

# Create a second list by adding the afternoon students to a **copy** of `morning` using `extend()`
morning_copy = morning.copy()
morning_copy.extend(afternoon)
print(morning_copy)

# Using a list comprehension, create a list of the numbers from `1` to `20` that are **odd**, each one squared
odd_squares = [x**2 for x in range(1, 21) if x % 2 == 1]
print(odd_squares)

# Using a list comprehension with a conditional expression, create a list of the numbers from `1` to `10` where even numbers are replaced by `"even"` and odd numbers stay as they are
transformed_numbers = ["even" if x % 2 == 0 else x for x in range(1, 11)]
print(transformed_numbers)
````

1. What is the difference between `+` and `extend()`?
-- "+" creates a new list by concatenating two lists, while `extend()` modifies an existing list by adding elements from another list.

2. Where does the `if` go in a comprehension that *filters*, and where does it go in one that *transforms*?
-- In a comprehension that filters, the `if` condition goes at the end of the expression, after the `for` loop. In a comprehension that transforms, the `if` condition is part of a conditional expression that determines what value to include in the new list.

## **Problem B3 (Set operations)**

Given:

```python
python_course = {"Ali", "Sara", "Lina", "Omar", "Noor"}
sql_course = {"Lina", "Omar", "Yara", "Zain"}
```

Print:
- students attending **both** courses
- students attending **at least one** course
- students attending **only** the Python course
- students attending **exactly one** course (not both)
- `True` or `False`: is `{"Lina", "Omar"}` a subset of `python_course`?
- `True` or `False`: do the two courses have **no** students in common?

### Reflection
- Which set operation answers each question above? Match them.

````python
python_course = {"Ali", "Sara", "Lina", "Omar", "Noor"}
sql_course = {"Lina", "Omar", "Yara", "Zain"}

# Find students enrolled in both courses
both_courses = python_course.intersection(sql_course)
print(both_courses) 

# Find students enrolled in at least one course
at_least_one = python_course.union(sql_course)
print(at_least_one)  

# Find students enrolled in the Python course but not in the SQL course
python_only = python_course.difference(sql_course)  
print(python_only) 

# Find Students attending only exactly one course
exactly_one = python_course.symmetric_difference(sql_course)
print(exactly_one)  

# True or False: Is the student {"Lina", "Omar"} subset of python_course?
is_subset = {"Lina", "Omar"}.issubset(python_course)
print(is_subset)  

# Do the two courses have any students in common?
have_common_students = bool(python_course.intersection(sql_course)) 
print(have_common_students) 
````

## **Problem C2 (Keys, values and items)**

Given:

```python
prices = {"apple": 1.2, "bread": 2.5, "cheese": 6.8, "milk": 1.1}
```

Tasks:
- Print all keys, then all values
- Print each item as `apple costs 1.2 EUR` (one per line) using a `for` loop over `.items()`
- Calculate and print the **total price** of all items
- Find and print the **most expensive** item and its price, hint use max(prices, key=prices.get)
- Update the price of `milk` to `1.3` using `update()`

### Reflection
- What does `.items()` return, and why is it convenient in a `for` loop?

````python
prices = {"apple": 1.2, "bread": 2.5, "cheese": 6.8, "milk": 1.1}

# Print all keys, then all values
print(prices.keys())
print(prices.values())

# Print each item as `apple costs 1.2 EUR` (one per line) using a `for` loop over `.items()`
for item, price in prices.items():
    print(f"{item} costs {price} EUR")

# Calculate and print the **total price** of all items
print(sum(prices.values()))

# Find and print the **most expensive** item and its price, hint use max(prices, key=prices.get)
most_expensive_item = max(prices, key=prices.get)
print(f"The most expensive item is {most_expensive_item} with a price of {prices[most_expensive_item]} EUR")

# Update the price of `milk` to `1.3` using `update()`
update_dict = {"milk": 1.3}
prices.update(update_dict)
print(prices)
````

1. What does ".items()" return, and why is it convenient in a "for" loop?
-- ".items()" returns a view object that displays a list of a dictionary's key-value tuple pairs. It is convenient in a "for" loop because it allows you to iterate over both the keys and values of the dictionary simultaneously.

## **Problem C4 (Counting with dictionaries)**

Given the sentence:

```python
text = "the cat saw the dog and the dog saw the bird"
```

Tasks:
- Split the sentence into words
- Using a `for` loop, build a dictionary `word_count` where each key is a word and each value is how many times it appears
- Print the dictionary
- Print the word that appears most often

Expected result (order may vary): `{'the': 4, 'cat': 1, 'saw': 2, 'dog': 2, 'and': 1, 'bird': 1}`

### Reflection
- Why is a dictionary a better fit than a list for counting?
- What do you need to check before increasing the count of a word?

````python
text = "the cat saw the dog and the dog saw the bird"

# Split the sentence into words
words = text.split()
print(words)

# Using a `for` loop, build a dictionary `word_count` where each key is a word and each value is how many times it appears
word_count = {}
for word in words:
    if word in word_count:
        word_count[word] +=1
    else:
        word_count[word] = 1
    
# Print the dictionary
print(word_count)

# Print the word that appears most often
most_common_word = max(word_count, key=word_count.get)
print(most_common_word)
````

1.  Why is a dictionary a better fit than a list for counting?
-- A dictionary allows for efficient lookup and modification of counts using the word as a key, whereas a list would require searching through the entire list to find the word and update its count.
2. What do you need to check before increasing the count of a word?
-- You need to check if the word is already a key in the dictionary. If it is, you increment its value; if not, you add it as a new key with an initial value of 1.

## **Problem D1 (List of dictionaries)**

Given:

```python
students = [
    {"name": "Anna", "grade": 1.7},
    {"name": "Ben", "grade": 2.3},
    {"name": "Clara", "grade": 1.3}
]
```

Tasks:
- Print the name of the second student
- Change Ben's grade to `2.0`
- Add a new student `{"name": "David", "grade": 2.7}` to the list
- Using a `for` loop, print every student as `Anna: 1.7`
- Using a list comprehension, create a list with only the **names**

### Reflection
- How do you read `students[0]["name"]` step by step?

````python
students = [
    {"name": "Anna", "grade": 1.7},
    {"name": "Ben", "grade": 2.3},
    {"name": "Clara", "grade": 1.3}
]

# Print the name of the second student
print(students[1]["name"])

# Change Ben's grade to `2.0`
students[1]["grade"] = 2.0
print(students)

# Add a new student `{"name": "David", "grade": 2.7}` to the list
students.append({"name": "David", "grade": 2.7})
print(students)

# Using a `for` loop, print every student as `Anna: 1.7`
for student in students:
    print(f"{student['name']}: {student['grade']}")
print(students)

# Using a list comprehension, create a list with only the **names**
names = [student['name'] for student in students]
print(names)
````

-- "["name"]" accesses the value associated with the key "name" in that dictionary.

## **Problem D3 (Flattening a nested list)**

Given:

```python
matrix = [
    [9, 3, 8, 3],
    [4, 5, 2, 8],
    [6, 4, 3, 1],
    [1, 0, 4, 5]
]
```

Tasks:
- Using a list comprehension, flatten the matrix into one list
- Create a list that contains only the **even** values of the flattened list
- Print the number of **unique** values in the matrix
- Print the sum of each row as a list (e.g. `[23, 19, 14, 10]`)

### Reflection
- In `[item for row in matrix for item in row]`, which loop is the outer one?

````python
matrix = [
    [9, 3, 8, 3],
    [4, 5, 2, 8],
    [6, 4, 3, 1],
    [1, 0, 4, 5]
]

# Using a list comprehension, flatten the matrix into one list
Flattened_list = [item for row in matrix for item in row]
print(Flattened_list)

# Create a list that contains only the **even** values of the flattened list
Even_values = [i for i in Flattened_list if i%2 == 0]
print(Even_values)

# Print the number of **unique** values in the matrix
unique_values = len(set(Flattened_list))
print(unique_values)

# Print the sum of each row as a list (e.g. `[23, 19, 14, 10]`)
row_sums = [sum(row) for row in matrix]
print(row_sums)
````

1. In "[item for row in matrix for item in row]", which loop is the outer one?
-- The outer loop is "for row in matrix", and the inner loop is "for item in row". This means that for each row in the matrix, it iterates through each item in that row.

## **Problem D4 (Mixed structures)**

Given an inventory where each category maps to a list of `(item, price)` tuples:

```python
inventory = {
    "fruit": [("apple", 1.2), ("mango", 2.5), ("lime", 0.4)],
    "dairy": [("milk", 1.1), ("cheese", 6.8)],
    "bakery": [("bread", 2.5), ("croissant", 1.9), ("bagel", 1.4)]
}
```

Tasks:
- Print every item name in the `"dairy"` category
- Print the **total price** of each category
- Find and print the **cheapest item overall** (name and price)
- Create a dictionary that maps each category to the **number of items** it contains

### Reflection
- Why is a tuple a good choice for each `(item, price)` pair?

````python
inventory = {
    "fruit": [("apple", 1.2), ("mango", 2.5), ("lime", 0.4)],
    "dairy": [("milk", 1.1), ("cheese", 6.8)],
    "bakery": [("bread", 2.5), ("croissant", 1.9), ("bagel", 1.4)]
}

# Print every item name in the `"dairy"` category
print([item[0] for item in inventory["dairy"]])

# Print the **total price** of each category
for category, items in inventory.items():
    total_price = sum(price for _, price in items)
    print(f"Total price of {category}: {total_price}")

# Find and print the **cheapest item overall** (name and price)
cheapest_item = min((item, price) for category, items in inventory.items() for item, price in items)
print(f"Cheapest item: {cheapest_item[0]}, Price: {cheapest_item[1]}")

# Create a dictionary that maps each category to the **number of items** it contains
category_counts = {category: len(items) for category, items in inventory.items()}
print(category_counts)
````

1. Why is a tuple a good choice for each "(item, price)" pair?
-- A tuple is a good choice for each "(item, price)" pair because tuples are immutable, meaning that once they are created, their values cannot be changed. This immutability ensures that the item and its corresponding price remain consistent and cannot be accidentally modified, which is important for maintaining data integrity in an inventory system.
